```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter, CharacterTextSplitter
from langchain_ollama import OllamaEmbeddings
from langchain_community.vectorstores import FAISS

#Ingestion of PDF document
loader = PyPDFLoader("Prasad_DE.pdf")
document = loader.load()
print((document))


#Chunking of document into smaller pieces
text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=0)
chunk = text_splitter.split_documents(document)
print((chunk))

#Embedding of chunks
embeddings = OllamaEmbeddings(model="nomic-embed-text")
emedded_chunks = embeddings.embed_documents([c.page_content for c in chunk])
emedded_chunks

## embed_query is used for embedding a string to vectors, where as embed_documents is used for embedding a list of strings documents to vectors.
#Creation of FAISS vector store
db = FAISS.from_documents(chunk, embeddings)  # embedding Metatdata + page content 
db
```

```python
#Similarity Search in vector DB directly using query string
query = "What is the Skillset of Prasad?"

results = db.similarity_search(query)
results

# similarity search via retriever interface
retriever = db.as_retriever()
results = retriever.invoke(query)
results

```
