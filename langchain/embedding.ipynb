```python
from langchain_openai import OpenAIEmbeddings

openai_embeddings = OpenAIEmbeddings(model="text-embedding-3-large")

text= "A vector database stores high-dimensional embeddings representing the semantic meaning of data and enables similarity-based retrieval, allowing applications such as RAG systems to find contextually relevant information rather than relying only on exact keyword matches."

embeddings = openai_embeddings.embed_query(text)

print(embeddings)
```

```python

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter  

loader = PyPDFLoader("Prasad_DE.pdf")
document = loader.load()

text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
final_doc = text_splitter.split_documents(document)
final_doc
```

```python
from langchain_community.vectorstores import Chroma

db = Chroma.from_documents(final_doc, openai_embeddings)

db
```
