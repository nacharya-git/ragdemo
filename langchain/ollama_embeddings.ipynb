```python
from langchain_ollama import OllamaEmbeddings

embeddings = OllamaEmbeddings(
    model="nomic-embed-text",
    base_url="http://localhost:11434"
)

text = "A vector database stores high-dimensional embeddings representing the semantic meaning of data and enables similarity-based retrieval, allowing applications such as RAG systems to find contextually relevant information rather than relying only on exact keyword matches."

result = embeddings.embed_query(text)


print(result)

```
